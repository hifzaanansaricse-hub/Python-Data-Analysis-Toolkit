```python
# Subplot - A subplot allows you to display multiple graphs (plots) in a single figure window.

# Why use it?

# Compare multiple datasets side by side.
# Save space by showing several charts in one figure.
# Create dashboards and reports.
```

```python
#plt.subplot(nrows,ncol,index)

import matplotlib.pyplot as plt 

x = [1,2,3,4]
y = [10,20,15,25]

plt.subplot(1,2,1) # 1st Row, 2nd Column, 1st Subplot
plt.plot(x,y)

plt.subplot(1,2,2) # 1st row , 2nd column, 2nd Subplot
plt.bar(x,y)

plt.title("Bar Chart")
plt.show()
```

```python
import matplotlib.pyplot as plt 

#fig,ax =plt.subplots(nrows,ncol,figsize =(width,height))

fig, ax = plt.subplots(1,2,figsize=(10,5))

x = [1,2,3,4]
y = [10,20,15,25]

ax[0].plot(x,y)
ax[0].set_title("Line Plot")

ax[1].bar(x,y)
ax[1].set_title("Bar Chart")
plt.tight_layout()
plt.show()
```
