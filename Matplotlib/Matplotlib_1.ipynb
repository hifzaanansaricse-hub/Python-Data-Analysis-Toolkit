```python
# Data Visualization

# Data visualization is the process of presenting data in a visual form, such as charts, graphs, maps, dashboards, and diagrams, to make information easier to understand and analyze.'''

# Advantages of Data Visualization

# Easy to understand data - Converts complex data into simple visual forms.

# Identifies trends and patterns - Helps quickly spot increases, decreases, and relationships.

# Improves decision-making - Provides clear insights for better business decisions.

# Saves time - Information can be understood faster than reading large tables.

# Detects errors and outliers - Unusual values and mistakes become easy to identify.
```

```python
import matplotlib.pyplot as plt

x = [1,2,3,4] # x - Horizontal axis
y = [60,55,65,60] # y - Vertical axis

plt.plot(x,y) # plt.plot() is a Matplotlib function used to create a line graph by plotting data points on the x-axis and y-axis.
plt.show()
```

```python
# Short Definitions
# Data Point: A single value in a dataset.
# X-axis: Horizontal axis of a graph.
# Y-axis: Vertical axis of a graph.
# Figure: The entire plotting window or canvas.
# Axes: The area where data is plotted.
# Plot: A visual representation of data.
# Marker: A symbol used to represent a data point.
# Line Style: The appearance of a line (solid, dashed, dotted).
# Legend: A box that explains graph elements.
# Label: Text describing an axis or plot.
# Title: The heading of a graph.
# Grid: Reference lines that make graphs easier to read.
# Function: A reusable block of code that performs a task.
# Method: A function associated with an object.
# Parameters: Variables listed in a function definition.
# Keyword Arguments: Arguments passed using name=value.
# Object-Oriented API: A way of creating plots using objects like Figure and Axes.
# DPI (Dots Per Inch): Measures image resolution.
# Backend: The system Matplotlib uses to display or save plots.
```

```python
import matplotlib.pyplot as plt 

x = ["Mon","Tue","Wed","Thurs","Fri"]
y = [ 10,15,7,20,12 ]

plt.plot(x,y)

plt.title("Bakery Sales This Week")
plt.xlabel("Days of the week")
plt.ylabel("Sales per day")
plt.show()
```
