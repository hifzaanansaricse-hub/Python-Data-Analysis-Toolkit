```python
# Merge and Joining 

# Syntax : pd.merge(d1,d2,on="Column_Name",how ="Type of join")
# d1 = First data frame
# d2 = Second Data Frame
# on = Name of that column which is common in both data frame
# how = Name of that operation
```

```python

""" 1. Inner Join
Keeps only the rows where the key (e.g., Customer_Id) exists in both DataFrames.
Result = intersection of both tables.
Example: Only customers who have placed orders."""

''' 2. Left Join
Keeps all rows from the left DataFrame (df_customers).
If there's a match in the right DataFrame (df_orders), it adds the data; otherwise, fills with NaN.
Example: All customers, with order info if available.'''

''' 3. Right Join
Keeps all rows from the right DataFrame (df_orders).
If there's a match in the left DataFrame (df_customers), it adds the data; otherwise, fills with NaN.
Example: All orders, with customer info if available.'''

'''4. Outer Join
Keeps all rows from both DataFrames.
Where there's no match, fills with NaN.
Example: All customers and all orders, even if some don't match.'''


import pandas as pd

# Customer DataFrame
df_customers = pd.DataFrame({
    "Customer_Id": [1, 2, 3, 4],
    "Name": ["Ramesh", "Suresh", "Kalpesh", "Mahesh"]
})

# Order DataFrame
df_orders = pd.DataFrame({
    "Customer_Id": [2, 3, 5],
    "OrderAmount": [250, 450, 350]
})

print("Customers DataFrame")
print(df_customers)
print("\nOrders DataFrame")
print(df_orders)
print("\n")

# 1. Inner Join
df_inner = pd.merge(df_customers, df_orders, on="Customer_Id", how="inner")
print(" Inner Join (only matching rows)")
print(df_inner)
print("\n")

# 2. Left Join
df_left = pd.merge(df_customers, df_orders, on="Customer_Id", how="left")
print(" Left Join (all customers, matching orders if available)")
print(df_left)
print("\n")

# 3. Right Join
df_right = pd.merge(df_customers, df_orders, on="Customer_Id", how="right")
print(" Right Join (all orders, matching customers if available)")
print(df_right)
print("\n")

# 4. Outer Join
df_outer = pd.merge(df_customers, df_orders, on="Customer_Id", how="outer")
print("- Outer Join (all rows from both, NaN where no match)")
print(df_outer)
```
