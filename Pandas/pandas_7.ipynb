```python
# Concatination

# In Pandas, concatenation means joining DataFrames together. It can be done vertically (row‑wise) or horizontally (column‑wise) depending on the axis parameter.

# Vertical Concatenation (Row‑wise)
# Stacks DataFrames on top of each other (adds more rows).
# Use axis=0 (default).
# Columns must match; otherwise, Pandas fills missing values with NaN.

import pandas as pd

df1 = pd.DataFrame({"Name":["Amit","Neha"], "Age":[28,32]})
df2 = pd.DataFrame({"Name":["Ravi","Karan"], "Age":[41,36]})

df_vertical = pd.concat([df1, df2], axis=0)
print(df_vertical)
print("\n")

# Horizontal Concatenation (Column‑wise)
# Joins DataFrames side by side (adds more columns).
# Use axis=1.
# Rows must align by index; otherwise, Pandas fills missing values with NaN.

df1 = pd.DataFrame({"Name":["Amit","Neha"], "Age":[28,32]})
df2 = pd.DataFrame({"Salary":[45000,52000], "Performance":[4,5]})

df_horizontal = pd.concat([df1, df2], axis=1)
print(df_horizontal)

# Vertical Concatenation (axis=0) → adds rows.

# Horizontal Concatenation (axis=1) → adds columns.
```
