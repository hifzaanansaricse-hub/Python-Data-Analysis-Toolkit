```python
import pandas as pd 

data = {
    "Name":["Ram","Shyam","Ravi"],
    "Age":[10,20,30],
    "City":["Nagpur","Mumbai","Delhi"]
}

df = pd.DataFrame(data)
print(df)

# For saving this data in your csv file.
# If you don't want index number so use index=false
df.to_csv("output.csv",index=False)

# For saving this data in your excel file.
df.to_excel("output.xlsx",index=False)

# For saving this data in your json file.
df.to_json("output.json",index=False)

```

```python
# Describe - The .describe() method gives summary statistics for numeric columns in your DataFrame. It quickly shows the distribution of values (count, mean, min, max, etc.), which helps you understand your dataset at a glance.


import pandas as pd

data = {
    "Name": ["Amit","Neha","Ravi","Priya","Karan","Meena","Suresh","Anita",
             "Vikram","Kavita","Arjun","Sunita","Deepak","Shweta","Manoj","Rekha"],
    "Age": [28,32,41,25,36,29,45,31,39,27,34,30,42,26,37,33],
    "Salary": [45000,52000,61000,38000,57000,49000,72000,51000,
               68000,42000,56000,50000,70000,39000,60000,54000],
    "Performance": [4,5,3,2,4,3,5,4,3,2,4,3,5,2,4,3]
}

df = pd.DataFrame(data)
print("Normal data set")
print(df) 


print("\n\n")

print("Descriptive Statistics")
print(df.describe())



# Output -

# Statistic    Meaning                                Example of output
# --------------------------------------------------------------------------------
# count    |    Number of non-missing values           Age → 16 (all rows have values)
# mean     |    Average value                          Age → 33.43 years, Salary → 54,000
# std      |    Standard deviation (spread of data)    Salary → 10,437 (variation around mean)
# min      |    Smallest value                         Age → 25, Salary → 38,000
# 25%      |    1st quartile (25% below this)          Age → 28.75, Salary → 48,000
# 50%      |    Median (middle value)                  Age → 32.5, Salary → 53,000
# 75%      |    3rd quartile (75% below this)          Age → 37.5, Salary → 60,250
# max      |    Largest value                          Age → 45, Salary → 72,000
```
