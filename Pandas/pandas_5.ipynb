```python
# Sorting

# Sorting means arranging data in order based on column values.

# sort_values(by="Column") → Sort rows by a specific column.

# sort_values(by=["Col1","Col2"]) → Sort by multiple columns.
```

```python
import pandas as pd

data ={
    "Name":["Arun","Varun","karun"],
    "Age":[78,56,45],
    "Salary":[10000,20000,30000]
}
df = pd.DataFrame(data)
print(df)
print("\n")

# inplace=True means: apply the changes directly to the original DataFrame without creating a new one.

# inplace=False (default) means: return a new DataFrame with the changes, while the original DataFrame stays unchanged.

df.sort_values(by="Age",ascending=False,inplace=True)
print('Sorted Age by Decending')
print("\n")
print(df)

print("\n")
df.sort_values(by="Age",ascending=True,inplace=True)
print('Sorted Age by Asecending')
print("\n")
print(df)
```

```python
import pandas as pd

data ={
    "Name":["Arun","Varun","karun"],
    "Age":[23,56,45],
    "Salary":[10000,20000,30000]
}
df = pd.DataFrame(data)
print(df)
print("\n")

df.sort_values(by=["Age","Salary"],ascending=False,inplace=True)
print("Sorted Age by Desending")
print(df)
print("\n")

# Multiple Columns Operation.

df.sort_values(by=["Age","Salary"],ascending=[True,False],inplace=True)
print("Sorted Age by Desending With Salary")
print(df)
print("\n")
```

```python
# In Pandas, Aggregation means summarizing data by applying statistical or mathematical functions to groups of rows or entire columns. It’s used to condense large datasets into meaningful values like totals, averages, counts, or maximums.

# sum() → Total of values

# mean() → Average

# count() → Number of items

# min() / max() → Smallest / largest value

# median() → Middle value
```

```python
import pandas as pd 
data = {
    "Department":["HR","HR","IT","IT","Finance","Finance"],
    "Salary":[40000,50000,60000,90000,30000,54000]
}

df = pd.DataFrame(data)

# Aggregation : Average salary per department

result = df.groupby("Department")["Salary"].mean()
print(result)
```

```python
# Summary Statistics 

import pandas as pd

data = {
    "Name": [
        "Arun","Varun","Karun","Meena","Suresh","Anita","Vikram","Kavita",
        "Rohan","Neha","Pooja","Deepak","Shweta","Manoj","Rekha","Rajesh",
        "Anjali","Kiran","Priya","Amit"
    ],
    "Age": [
        28,34,22,29,45,31,39,27,
        30,32,26,41,25,37,33,40,
        24,36,29,38
    ],
    "Salary": [
        10000,20000,30000,25000,40000,35000,45000,28000,
        32000,37000,26000,50000,22000,42000,39000,47000,
        24000,41000,33000,44000
    ]
}

df = pd.DataFrame(data)
print(df)


print("\n")

avg_salary = df["Salary"].mean()
print("The average salary is : ",avg_salary)

#GroupBy Method

grouped = df.groupby("Age")["Salary"].sum()
print(grouped)

print("\n")

grouped_2 = df.groupby(["Age","Name"])["Salary"].sum()
print(grouped_2)
```
