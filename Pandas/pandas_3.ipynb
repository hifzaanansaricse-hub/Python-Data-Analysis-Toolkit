```python
import pandas as pd 

data = {
    "Name": ["Amit","Neha","Ravi","Priya","Karan","Meena","Suresh","Anita",
             "Vikram","Kavita","Arjun","Sunita","Deepak","Shweta","Manoj","Rekha"],
    "Age": [28,32,41,25,36,29,45,31,39,27,34,30,42,26,37,33],
    "Salary": [45000,52000,61000,38000,57000,49000,72000,51000,
               68000,42000,56000,50000,70000,39000,60000,54000],
    "Performance": [4,5,3,2,4,3,5,4,3,2,4,3,5,2,4,3]
}


df =pd.DataFrame(data)
print(df)

# Selecting single columns
print("\n\n")
print("A Single Column\n")
print(df["Name"])

# Selecting Multiple Columns
print("\n\n")
print("Name with salary\n")
subset = df[["Name","Salary"]]
print(subset)

# Age with salary
print("\n\n")
print("Age with salary \n")
subset_1 =df[["Age","Salary"]]
print(subset_1)

# Name and salary decending order
print("\n\n Name and Salary decending order")
new_df = df[["Name", "Salary"]].sort_values(by="Salary", ascending=False).reset_index(drop=True)
print(new_df)
```

```python
import pandas as pd 

data = {
    "Name": ["Amit","Neha","Ravi","Priya","Karan","Meena","Suresh","Anita",
             "Vikram","Kavita","Arjun","Sunita","Deepak","Shweta","Manoj","Rekha"],
    "Age": [28,32,41,25,36,29,45,31,39,27,34,30,42,26,37,33],
    "Salary": [45000,52000,61000,38000,57000,49000,72000,51000,
               68000,42000,56000,50000,70000,39000,60000,54000],
    "Performance": [4,5,3,2,4,3,5,4,3,2,4,3,5,2,4,3]
}

df = pd.DataFrame(data)

# Single row filtering.

high_salary = df[df["Salary"]>50000]
print("The person which salary is above 50k\n")
print(high_salary)

# Multiple Condition 

# Using AND Condition

print("\n\n")
print("Mulitple Condition\n")
print("The list of person which age is greater than 30 and the salary is greater than 50k\n")

filtered = df[(df["Age"]>30) & (df["Salary"]>50000)]
print(filtered)

# Using OR Condition 
print("\n\n")
filtered_or = df[(df["Age"] > 35) | (df["Performance"] > 90)]
print("Employee older than 35 OR performance score > 90\n")
print(filtered_or)

```
