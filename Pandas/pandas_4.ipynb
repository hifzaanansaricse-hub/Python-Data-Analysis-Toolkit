```python
import pandas as pd

# Data with some missing values
data = {
    "Name": ["Amit","Neha","Ravi",None,"Karan","Meena","Suresh","Anita",
             "Vikram","Kavita",None,None,"Deepak","Shweta","Manoj","Rekha"],
    "Age": [28,32,41,25,36,29,45,31,39,27,34,30,42,26,37,33],
    "Salary": [45000,52000,61000,38000,57000,49000,72000,51000,
               68000,42000,None,50000,None,39000,60000,54000],
    "Performance": [4,5,3,2,4,3,None,4,3,2,4,3,None,2,4,3]
}

# Create DataFrame
df = pd.DataFrame(data)
print("\nOriginal Data Set\n")
print(df)

# Fill only numeric columns with 0
df[["Age","Salary","Performance"]] = df[["Age","Salary","Performance"]].fillna(0)

print("\nCleaned Data Set (NaN replaced with 0 in numeric columns)\n")
print(df)

```

```python
# Interpulation - Fill the estimated value in place of missing values.
# interpolate() only works on numeric columns.
import pandas as pd

data = {
    "Name": ["Amit","Neha","Ravi","Shyam","Karan","Meena","Suresh","Anita",
             "Vikram","Kavita","Rakhi","shamiya","Deepak","Shweta","Manoj","Rekha"],
    "Age": [28,None,41,25,36,29,45,31,39,27,34,30,42,26,37,33],
    "Salary": [45000,52000,61000,38000,57000,49000,72000,51000,
               68000,42000,None,50000,None,39000,60000,54000],
    "Performance": [4,5,3,2,4,3,None,4,3,2,4,3,None,2,4,3]
}

# Create DataFrame
df = pd.DataFrame(data)
print("\nOriginal Data Set\n")
print(df)

# Interpolate only numeric columns
df[["Age","Salary","Performance"]] = df[["Age","Salary","Performance"]].interpolate(method="linear", axis=0)

print("\nAfter Linear Interpolation\n")
print(df)

```

```python
# Pandas Methods Overview

# 1. Handling Missing Data

# fillna(value) → Replace missing values (NaN) with a given value.
# dropna() → Remove rows or columns with missing values.
# isna() / notna() → Check for missing values.
# interpolate(method="linear") → Fill missing values using linear interpolation.
# Other methods: "polynomial", "spline", "time".


# 2. Axis Operations


# sum(axis=0) → Column‑wise sum (default).
# sum(axis=1) → Row‑wise sum.
# mean(axis=0/1) → Average across rows or columns.
# apply(func, axis=0/1) → Apply a custom function row‑wise or column‑wise.
# concat([df1, df2], axis=1) → Combine along columns.
# axis=0 means operate down columns.
# axis=1 means operate across rows.


# 3. Time and Date Methods

# pd.to_datetime() → Convert strings to datetime objects.
# pd.date_range(start, periods, freq) → Generate a sequence of dates.
# resample("M").mean() → Resample time series (e.g., monthly average).
# Timestamp → Represents a single date/time.
# Timedelta → Represents a duration.
# dt.year, dt.month, dt.day, dt.hour → Extract parts of a datetime.
# tz_localize() / tz_convert() → Handle time zones.
# rolling(window) → Moving averages over time.


# 4. General DataFrame Methods


# head() / tail() → Preview rows.
# describe() → Summary statistics.
# groupby() → Group and aggregate data.
# merge() / join() → Combine datasets.
# pivot_table() → Create pivot tables.
# sort_values() → Sort by column values.
# astype() → Change data type of a column.
```
