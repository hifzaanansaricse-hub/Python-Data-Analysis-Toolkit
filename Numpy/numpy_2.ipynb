```python
# Zero-array function - In which we can put all array elements = zero. 

import numpy as np

print("The 1-D array")
zeroes_array = np.zeros(9)
print(zeroes_array)

print('\n')

print("The 2-D arry")
zeroes_array1 = np.zeros((3,3))
print(zeroes_array1)
```

```python
# One-array function - In which we can put all array elements  = One.

import numpy as np 

Ones_array = np.ones(9)
print("1*1: \n",Ones_array)

print("\n")

Ones_array2= np.ones((2,3))
print("2*3: \n",Ones_array2)

print("\n")

Ones_array3= np.ones((3,3))
print("3*3 : \n",Ones_array3)
```

```python
# For fill particular specific value - Used Full funtion 

import numpy as np 
Full_array = np.full((2,2),8)
print(Full_array)

print("\n")

Full_array2 = np.full((4,4),8)
print(Full_array2)

print("\n")
Full_array3 = np.full((6,6),6)
print(Full_array3)
```

```python
# Creating Sequences of number in numpy . 
# Use - Arange()
# Syntax - arange(start,stop,step)

# Example - print the table of number 5.

import numpy as np 
table = np.arange(5,55,5)

print("The Horizontally : ",table)
print('\n')

print( "The vertically : ",*table,sep="\n") # Point to be learn.
```

```python
# Creating Identity Matrix . 
# Use eye(size) function. 

import numpy as np 

Identity = np.eye(4)
Indentity2 = np.eye(3)
print(Identity)

print("\n")

print(Indentity2)
```
