```python
# Capstone Project

import pandas as pd          # Import pandas for data analysis
import numpy as np           # Import numpy for numerical operations

# Load the raw dataset
df = pd.read_csv(r"D:\N-P-M\Numpy\indian_employee_raw_data_500_rows.csv")

# Handle the Missing value.

df["Salary (INR)"].fillna(df["Salary (INR)"].mean(), inplace=True)  
# Replace missing salaries with the average salary

df["Performance_Rating"].fillna(df["Performance_Rating"].median(), inplace=True)  
# Replace missing performance ratings with the median rating

df.replace([np.inf, -np.inf], np.nan, inplace=True)  
# Convert infinite values to NaN (missing)

df.fillna(df.select_dtypes(include=[np.number]).mean(), inplace=True)  
# Fill remaining numeric NaN values with their column means

df.drop_duplicates(inplace=True)  
# Remove duplicate rows

df["Salary (INR)"] = np.where(df["Salary (INR)"] < 0, df["Salary (INR)"].mean(), df["Salary (INR)"])  
# Replace negative salaries with the average salary

salary_mean = df["Salary (INR)"].mean()  
# Calculate mean salary

salary_std = df["Salary (INR)"].std()  
# Calculate standard deviation of salary

lower_bound = salary_mean - (3 * salary_std)  
# Define lower limit for outlier detection (3 std below mean)

upper_bound = salary_mean + (3 * salary_std)  
# Define upper limit for outlier detection (3 std above mean)

df = df[(df["Salary (INR)"] >= lower_bound) & (df["Salary (INR)"] <= upper_bound)]  
# Keep only rows where salary is within the normal range (remove outliers)

df.to_csv("Cleaned_Indian_Employee.csv", index=False)  
# Save the cleaned dataset to a new CSV file

print("Data cleaning completed !")  
# Print confirmation message
```
