```python
# Broadcasting & Vectorization.
# Broadcasting - A way NumPy automatically stretches arrays of different shapes so they can work together in operations without loop.
# Vectorization - Performing operations on whole arrays at once instead of looping through elements one by one.
```

```python
# Normal python code 

prices = [100,200,300]
discount = 10

final_prices =[]

for price in prices:
    final_price = price - (price * discount/100)
    final_prices.append(final_price)

print(final_prices)    
```

```python
# With using Numpy.

import numpy as np 

prices = np.array([100,200,300])
discount =10 # Scaler(Single Value)

final_prices = prices - (prices * discount/100)

print(final_prices)
```

```python
# Handling Missing values in array.
# NaN - Not a Number.
# np.isnan() - Used for fnindig the missing value.
# np.nan-to-numc() -Used replace the value with missing value.
# np.isinf() - Used to find infinite value.
```

```python
import numpy as np 
array1 = np.array([1,2,np.nan,3,4,np.nan,5,6])
print(np.isnan(array1))

# Output - Boolean form 
```

```python
import numpy as np 
array1 = np.array([1,2,np.nan,3,4,np.nan,5,6])

array2 = np.nan_to_num(array1)
print("Without any particular value",array2)
print("\n")

# Give particular value like - 40

array2 = np.nan_to_num(array1,nan=40)
print("If put any particular value example -40 :",array2)
```

```python
# Checking the Infinite number 

import numpy as np 
array1 = ([1,2,np.inf,3,4,np.inf])

print(np.isinf(array1))
```

```python
# Replace any Infinte value.

import numpy as np 
array1 = ([1,2,np.inf,3,4,-np.inf])

print(np.nan_to_num(array1,posinf=8,neginf=9))
```
